```python
import matplotlib.pyplot as plt
import numpy as np


x=np.linspace(1,10,100)
y=np.sin(x)
plt.plot(x,y,label="sine wave",linestyle="--",color="blue",linewidth=2)
plt.title("Simple plot ")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
# scatter plot

x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y,color="red",marker="o",alpha=0.7,s=100)
plt.title("Scatter Plot ")
plt.xlabel("X-axis ")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ["A", "B", "C", "D"]
values = [10, 15, 7, 20]
plt.bar(categories,values,color=["red", "blue", "green", "purple"],edgecolor="black")
plt.title("Bar chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Histogram

data = np.random.randn(1000)

plt.hist(data,bins=30,alpha=0.7,color="green",edgecolor="black")
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd

# Simulating stock data
days = np.arange(1, 11)
prices = np.array([100, 102, 100, 107, 110, 111, 115, 118, 120, 118])

plt.plot(days, prices, marker="o", linestyle="-", color="green", label="Stock Price")
plt.title("Stock Market Trends")
plt.xlabel("Days")
plt.ylabel("Stock Price ($)")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data
np.random.seed(42)
data = np.random.randn(100)  # 100 random numbers from normal distribution

# Create boxplot
plt.boxplot(data)

plt.title("Basic Boxplot Example")
plt.xlabel("Data")
plt.ylabel("Values")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]

# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot in the first subplot
ax1.plot(x, y1, color='blue')
ax1.set_title("y = x^2")

# Plot in the second subplot
ax2.plot(x, y2, color='red')
ax2.set_title("y = x")

# Display the plots
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample dataset
X = np.array([1000, 1500, 2000, 2500, 3000]).reshape(-1, 1)  # Square feet
y = np.array([200000, 250000, 300000, 350000, 400000])  # Prices

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict
value=int(input("Enter the value"))
prediction = model.predict([[value]])
print(f"Predicted price for {value} sq ft: ${prediction[0]:.2f}")

# Plot
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Square Feet")
plt.ylabel("Price")
plt.show()
```
